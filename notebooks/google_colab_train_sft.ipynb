# XAI - Train SFT on Google Colab

Notebook nay train truc tiep tren Colab GPU, khong can thu muc `models/` local.

```python
# ====== 1) Cau hinh ======
REPO_URL = "https://github.com/Kunneaa/XAI.git"
BRANCH = "main"
WORKDIR = "/content/XAI"

# Chon model:
# - USE_LOCAL_MODEL=True: dung /content/XAI/models (file nam truc tiep trong models/)
# - USE_LOCAL_MODEL=False: tai model tu Hugging Face
USE_LOCAL_MODEL = False
LOCAL_MODEL_PATH = "models"
HF_MODEL_ID = "Qwen/Qwen2.5-1.5B-Instruct"

# Data + output
DATA_FILE = "training/competition_sft.jsonl"
OUTPUT_DIR = "training/checkpoints/logic_nl2fol"

# Hyperparameters toi uu cho T4
EPOCHS = 4
BATCH_SIZE = 2
GRAD_ACCUM = 8
MAX_LEN = 512
LR = 1e-5
EVAL_RATIO = 0.1
EVAL_STEPS = 20
EARLY_STOPPING_PATIENCE = 3
RESUME_FROM_CHECKPOINT = ""  # de trong de train lai tu dau
```

```python
# ====== 3) Clone repo (safe) ======
import os
import shutil

os.chdir("/content")
if os.path.exists(WORKDIR):
    shutil.rmtree(WORKDIR)

!git clone -b {BRANCH} {REPO_URL} {WORKDIR}
%cd {WORKDIR}
```

```python
# ====== 5) Tao data SFT ======
!python training/prepare_logic_sft_data.py
!python training/prepare_competition_sft_data.py
!ls -lh training/*.jsonl
```

```python
# ====== 6) Chon model argument ======
MODEL_ARG = LOCAL_MODEL_PATH if USE_LOCAL_MODEL else HF_MODEL_ID
print("MODEL_ARG =", MODEL_ARG)
```

```python
# ====== 7) Xoa checkpoint cu de train lai tu dau ======
import os
import shutil
if os.path.exists(OUTPUT_DIR):
    shutil.rmtree(OUTPUT_DIR)
print("Cleaned:", OUTPUT_DIR)
```

```python
# ====== 7) Train SFT ======
EXTRA_RESUME = f" --resume-from-checkpoint {RESUME_FROM_CHECKPOINT}" if RESUME_FROM_CHECKPOINT else ""
cmd = f"""python training/train_sft.py \
  --model {MODEL_ARG} \
  --data {DATA_FILE} \
  --out {OUTPUT_DIR} \
  --epochs {EPOCHS} \
  --bs {BATCH_SIZE} \
  --grad-accum {GRAD_ACCUM} \
  --max-len {MAX_LEN} \
  --lr {LR} \
  --eval-ratio {EVAL_RATIO} \
  --eval-steps {EVAL_STEPS} \
  --early-stopping-patience {EARLY_STOPPING_PATIENCE} \
  --chat-template""" + EXTRA_RESUME
print(cmd)
import os
os.system(cmd)
```
